```python
from contextlib import ExitStack
from pathlib import Path
from judge_predict import seed_everything, load_models, load_input, predict
seed_everything()
```

```python
if "session" in globals():
    session.close()
session = ExitStack()
try:
    models = load_models(session)
except BaseException:
    session.close()
    raise
```

```python
try:
    dataset = load_input()
except BaseException:
    session.close()
    raise
```

```python
try:
    predict(models, dataset, Path("/result/jupyter_submission.jsonl"))
finally:
    session.close()
```
